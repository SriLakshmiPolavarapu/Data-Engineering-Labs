```python
import pandas as pd
import random
import numpy as np
from faker import Faker

fake = Faker()
Faker.seed(42)
np.random.seed(42)

emp_df = pd.read_csv("employees.csv")
original_columns = emp_df.columns.tolist()

num_employees = 10000
languages = ["Spanish", "Mandarin", "Hindi", "French", "Korean", "Tagalog", "German", "Russian"]

h1b_country_dist = {
    "India": 0.72,
    "China": 0.13,
    "Canada": 0.05,
    "South Korea": 0.03,
    "Philippines": 0.03,
    "Taiwan": 0.02,
    "Mexico": 0.02
}

department_distribution = {
    "Legal": 0.05,
    "Marketing": 0.10,
    "Administrative": 0.10,
    "Operations": 0.20,
    "Sales": 0.10,
    "Finance": 0.05,
    "I/T": 0.10,
    "Product": 0.20,
    "Human Resource": 0.10
}

salary_ranges = {
    "Legal": (90000, 130000),
    "Marketing": (60000, 85000),
    "Administrative": (40000, 60000),
    "Operations": (50000, 75000),
    "Sales": (55000, 90000),
    "Finance": (70000, 110000),
    "I/T": (80000, 120000),
    "Product": (90000, 130000),
    "Human Resource": (50000, 80000)
}

job_titles = [
    "Software Engineer", "Data Scientist", "Product Manager", "Web Developer",
    "Systems Analyst", "Marketing Specialist", "HR Generalist", "Finance Analyst",
    "Operations Manager", "Legal Advisor", "Machine Learning Engineer", "Mobile Developer"
]

h1b_pool = []
for country, pct in h1b_country_dist.items():
    count = int(num_employees * 0.4 * pct)
    h1b_pool += [country] * count
h1b_pool += ["USA"] * (num_employees - len(h1b_pool))
random.shuffle(h1b_pool)

data = []
for i in range(num_employees):
    gender = "female" if i % 2 == 0 else "male"
    dept = random.choices(list(department_distribution.keys()), weights=department_distribution.values())[0]
    salary = round(np.random.uniform(*salary_ranges[dept]), 2)
    years_exp = random.randint(0, 30)
    age = random.randint(21, 60)
    languages_spoken = ", ".join(random.sample(languages, random.randint(0, 2)))
    citizenship = h1b_pool[i]
    ssn = fake.ssn()

    record = {
        "First Name": fake.first_name_female() if gender == "female" else fake.first_name_male(),
        "Last Name": fake.last_name(),
        "Email": fake.email(),
        "Phone": fake.phone_number(),
        "Gender": gender,
        "Age": age,
        "Job Title": random.choice(job_titles),
        "Years Of Experience": years_exp,
        "Salary": salary,
        "Department": dept,
        "SSN": ssn,
        "Languages Spoken": languages_spoken,
        "Citizenship": citizenship
    }
    data.append(record)

synth_df = pd.DataFrame(data)

print(synth_df.head(10))

```

```python

prtrb_df = emp_df.copy()

age_std = emp_df["Age"].std() * 0.1
salary_std = emp_df["Salary"].std() * 0.15
exp_std = emp_df["Years Of Experience"].std() * 0.2

np.random.seed(42)
prtrb_df["Age"] = (emp_df["Age"] + np.random.normal(0, age_std, len(emp_df))).round().clip(lower=18)
prtrb_df["Salary"] = (emp_df["Salary"] + np.random.normal(0, salary_std, len(emp_df))).round(2).clip(lower=0)
prtrb_df["Years Of Experience"] = (
    emp_df["Years Of Experience"] + np.random.normal(0, exp_std, len(emp_df))
).round().clip(lower=0)


print(prtrb_df.describe(include='all'))

```
